## Cosine Similarity
Cosine similarity measures how close two pieces of text are by finding the cosine of the angle between their vector representations, focuing on their direction rather than their magnitude

1. Score of 1: The angle is 0
degrees; the texts point in the exact same direction and are very similar.
2. Score of 0: The angle is 90 degrees; the texts are orthogonal (unrelated or share no common context).
3. Score of -1: The angle is 180 degrees; the texts point in opposite directions (rare in standard text matching).

```python
df = pd.DataFrame(faq_data)
```

```python
df.head()
```

```python
user = model.encode(input("What is ur query?: "))
faq = model.encode(df["Questions"].tolist())
```

```python
id = util.semantic_search(user,faq, top_k = 1)[0][0]["corpus_id"]
df["answer"][id]
```

```python
def chatbot():
  print("Chatbot: Hello! Ask a question or type exit to quit.")
  while True:
    user_query = input("You : ")
    if user_query.lower() in ["exit","quit"]:
      print("Chatbot: Goodbye!")
      break
    query_emb = model.encode(user_query)

    closest = util.semantic_search(query_emb,faq,top_k = 1)
    best_match_id = closest[0][0]["corpus_id"]
    best_answer = df.iloc[best_match_id]["answer"]

    print(f"Chatbot : {best_answer}")
```

```python
chatbot()
```
