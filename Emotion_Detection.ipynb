```python
from pysentimiento import create_analyzer
```

```python
emotion = create_analyzer(task = "emotion",lang = "en")
```

```python
emotion.predict("i feel delighted").output
```

```python
def emotion_detection(text):
  emotion = create_analyzer(task = "emotion",lang = "en")
  result = emotion.predict(text)
  return result.output
```

```python
emotion_detection("I am tired")
```

## Hate speech detection


```python
hate_speech = create_analyzer(task = "hate_speech",lang = "en")
```

```python
text = """You are the dumbest person i have ever met,
your ideas are trash and nobody wants to hear you speak"""
hate_speech.predict(text)
```
